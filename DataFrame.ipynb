# Data Loading, storage and File formats

Data Frame

```python

import pandas as pd
data={
    "Name":["John","Anna","Peter","karan"],
    "Age":[28,24,35,68],
    "City":["New York","USA","Paris","Landon"],
    "Grade":["A","B","C","A+"]
}
df=pd.DataFrame(data)
df
```

```python
print("Shape of the dataframe:",df.shape)
print("Head with 2 rows: \n",df.head(2))
print("Tail of 2 row:\n",df.tail(2))
print("Column:",df.columns.tolist())
print("Data Type:\n ",df.dtypes)
print("Data Information:\n ",df.info())
print("Statistic of DataFrame:\n ",df.describe())
```

#File Formats
1. csv
2. excel
3. json
4. parquet

csv

```python

#Load the dataset in the format
iris.to_csv("iris_csvoutput.csv",index=False)
# read the saved dataset
iris_csv=pd.read_csv("/content/iris_csvoutput.csv")
iris_csv
print(iris_csv.head())
```

excel

```python
iris.to_excel("iris_exceloutput.xlsx",index=False)
#Load
iris_excel =pd.read_excel("/content/iris_exceloutput.xlsx")
print(iris_excel.tail(10))
```

Json

```python
#Json file format
iris.to_json("iris_jsonoutput.json",orient="records",indent=4)
#read the file
iris_json=pd.read_json("/content/iris_jsonoutput.json")
iris_json.head(5)
```

Parquet

```python
#File format parquet
iris.to_parquet("iris_parquetoutput.parquet",index=False)
iris_parquet=pd.read_parquet("/content/iris_parquetoutput.parquet")
iris_parquet.head(3)
```

```python
#Compare the size
import os
files={
    "iris_csvoutput.csv",
    "iris_exceloutput.xlsx",
    "iris_jsonoutput.json",
    "iris_parquetoutput.parquet"
}
for file in files:
  size = os.path.getsize(file)
  print(f"{file}:{size} bytes")
```
