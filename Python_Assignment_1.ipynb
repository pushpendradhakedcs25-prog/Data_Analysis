#Topic 7: Functions & passing arguments

#Question 1

```python
def my_abs(val):
    if val < 0:
        return -val
    return val

print(my_abs(5))
print(my_abs(-10))
print(my_abs(0))
```

#Question 2

```python
##Prediction: It will raise a TypeError.
##Reasoning: Inside my_abs, Python attempts to evaluate val < 0.
##Comparing a string ("Hi") to an integer (0) using the < operator is not allowed in Python.
```

#Question 3

```python
def print_abs(val):
    if val < 0:
        print(-val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)



## printed for x: None
##Why: The function print_abs prints the absolute value to the console but does not contain a return statement. In Python, a function without an explicit return automatically returns None.
```

#Question 5
